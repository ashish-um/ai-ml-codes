## EXPERT SYSTEM

```python
class ExpertSystem:

    def __init__(self, facts, rules):
        self.facts = set(facts)
        self.rules = rules

    def _eval_condition(self, expr, known_facts):
        """Recursively evaluates nested conditions against known facts for Forward Chaining."""
        if isinstance(expr, str):
            return expr in known_facts

        operator, *operands = expr
        if operator == "AND":
            return all(self._eval_condition(op, known_facts) for op in operands)
        elif operator == "OR":
            return any(self._eval_condition(op, known_facts) for op in operands)
        else:
            raise ValueError(f"Invalid operator '{operator}'. Use 'AND' or 'OR'.")

    # Forward Chaining
    def forward_chaining(self, goal):
        known_facts = set(self.facts)
        inference_steps = []
        changed = True

        while changed:
            changed = False

            for rule_number, (condition, conclusion) in enumerate(self.rules, 1):
                if conclusion not in known_facts:
                    if self._eval_condition(condition, known_facts):
                        known_facts.add(conclusion)
                        inference_steps.append((rule_number, condition, conclusion))
                        changed = True

                        if conclusion == goal:
                            return True, inference_steps

        return goal in known_facts, inference_steps

    # Backward Chaining
    def backward_chaining(self, goal):
        visited = set()
        inference_steps = []

        def prove_condition(expr, depth):
            if isinstance(expr, str):
                return prove(expr, depth)

            operator, *operands = expr
            if operator == "AND":
                for op in operands:
                    if not prove_condition(op, depth):
                        return False
                return True

            elif operator == "OR":
                for op in operands:
                    if prove_condition(op, depth):
                        return True
                return False

            else:
                raise ValueError(f"Invalid operator '{operator}'. Use 'AND' or 'OR'.")

        def prove(target, depth=0):
            if target in self.facts:
                inference_steps.append(("FACT", target, depth))
                return True

            if target in visited:
                return False

            visited.add(target)

            for rule_number, (condition, conclusion) in enumerate(self.rules, 1):
                if conclusion == target:
                    if prove_condition(condition, depth + 1):
                        inference_steps.append(
                            ("RULE", rule_number, condition, target, depth)
                        )
                        return True

            return False

        result = prove(goal)
        return result, inference_steps
```

```python
def format_condition(expr):
    """Recursively formats nested condition trees into human-readable strings."""
    if isinstance(expr, str):
        return expr

    operator, *operands = expr
    formatted_operands = [format_condition(op) for op in operands]

    # Join operands with operator and wrap in parentheses
    return f"({f' {operator} '.join(formatted_operands)})"


def display_facts(facts):
    print("\nInitial Facts")
    print("-" * 60)

    for i, fact in enumerate(sorted(facts), 1):
        print(f"{i}. {fact}")

    print("-" * 60)


def display_rules(rules):
    print("\nRules")
    print("-" * 60)

    for i, (condition, conclusion) in enumerate(rules, 1):
        condition_text = format_condition(condition)

        # Strip outermost parentheses for cleaner top-level printing
        if condition_text.startswith("(") and condition_text.endswith(")"):
            condition_text = condition_text[1:-1]

        print(f"R{i}: IF {condition_text} THEN {conclusion}")

    print("-" * 60)


def display_goals(goals):
    print("\nPossible Goals")
    print("-" * 60)

    for i, goal in enumerate(goals, 1):
        print(f"{i}. {goal}")

    print("-" * 60)
```

```python
def format_condition(expr):
    """Recursively formats nested condition trees into human-readable strings."""
    if isinstance(expr, str):
        return expr

    operator, *operands = expr
    formatted_operands = [format_condition(op) for op in operands]
    return f"({f' {operator} '.join(formatted_operands)})"


def show_forward_result(system, goal):

    print("\n" + "=" * 70)
    print("FORWARD CHAINING")
    print("=" * 70)

    result, steps = system.forward_chaining(goal)

    print(f"\nGoal: {goal}")

    if steps:

        print("\nInference Steps:")

        for step_no, (rule_number, condition, conclusion) in enumerate(steps, 1):

            condition_text = format_condition(condition)

            # Strip outer parentheses for clean display
            if condition_text.startswith("(") and condition_text.endswith(")"):
                condition_text = condition_text[1:-1]

            print(
                f"{step_no}. "
                f"R{rule_number}: "
                f"{condition_text} "
                f"-> {conclusion}"
            )

    else:
        print("\nNo new facts were inferred.")

    print("\nFinal Inference:")

    if result:
        print(f"SUCCESS: '{goal}' can be inferred.")
    else:
        print(f"FAILURE: '{goal}' cannot be inferred.")

    print("=" * 70)
```

```python
def format_condition(expr):
    """Recursively formats nested condition trees into human-readable strings."""
    if isinstance(expr, str):
        return expr

    operator, *operands = expr
    formatted_operands = [format_condition(op) for op in operands]
    return f"({f' {operator} '.join(formatted_operands)})"


def show_backward_result(system, goal):

    print("\n" + "=" * 70)
    print("BACKWARD CHAINING")
    print("=" * 70)

    result, steps = system.backward_chaining(goal)

    print(f"\nGoal: {goal}")

    if result:

        print("\nInference Path:")

        for step in reversed(steps):

            if step[0] == "FACT":

                _, fact, depth = step

                print(
                    f"{'  ' * depth}FACT: {fact}"
                )

            elif step[0] == "RULE":

                _, rule_number, condition, conclusion, depth = step

                condition_text = format_condition(condition)

                # Strip outer parentheses for clean display
                if condition_text.startswith("(") and condition_text.endswith(")"):
                    condition_text = condition_text[1:-1]

                print(
                    f"{'  ' * depth}"
                    f"R{rule_number}: "
                    f"{condition_text} "
                    f"-> {conclusion}"
                )

        print("\n✓ SUCCESS")
        print(f"Goal '{goal}' has been proved.")

    else:

        print("\n✗ FAILURE")
        print(
            f"Goal '{goal}' cannot be proved "
            "from the available facts and rules."
        )

    print("=" * 70)
```

```python
# RUN SELECTED CASE

def run_case(case_number):

    case = cases[case_number]

    print("\n" + "=" * 70)
    print(f"CASE {case_number}: {case['name']}")
    print("=" * 70)

    # Display knowledge base
    display_facts(case["facts"])
    display_rules(case["rules"])

    system = ExpertSystem(
        case["facts"],
        case["rules"]
    )

    goal = select_goal(case["goals"])

    print(f"\nSelected Goal: {goal}")

    # Select inference method
    method = select_method()

    if method == 1:

        show_forward_result(system, goal)

    elif method == 2:

        show_backward_result(system, goal)

    elif method == 3:

        show_forward_result(system, goal)
        show_backward_result(system, goal)
```
