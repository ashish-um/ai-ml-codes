#### IMPORTING LIBRARIES FOR CHATBOT

```python
import tkinter as tk
from tkinter import scrolledtext
```

#### ENVIRONMENT

```python
class AdmissionChatbot:

    def __init__(self, root):

        self.root = root
        self.root.title("College Admission Chatbot")
        self.root.geometry("600x650")

        self.simple_agent = SimpleReflexAgent()
        self.model_agent = ModelBasedAgent()

        title_label = tk.Label(
            root,
            text="College Admission Assistant",
            font=("Arial", 18, "bold")
        )
        title_label.pack(pady=10)

        agent_label = tk.Label(
            root,
            text="Select Agent:",
            font=("Arial", 11)
        )
        agent_label.pack()

        self.agent_type = tk.StringVar()
        self.agent_type.set("Simple Reflex Agent")

        agent_menu = tk.OptionMenu(
            root,
            self.agent_type,
            "Simple Reflex Agent",
            "Model-Based Agent"
        )
        agent_menu.pack(pady=5)

        self.chat_area = scrolledtext.ScrolledText(
            root,
            wrap=tk.WORD,
            width=65,
            height=25,
            font=("Arial", 11)
        )
        self.chat_area.pack(
            padx=10,
            pady=10
        )

        self.user_input = tk.Entry(
            root,
            width=55,
            font=("Arial", 12)
        )
        self.user_input.pack(
            side=tk.LEFT,
            padx=(15, 5),
            pady=10
        )

        self.sensor = AdmissionSensor(self)

        self.actuator = AdmissionActuator(self)

        send_button = tk.Button(
            root,
            text="Send",
            command=self.send_message,
            width=10
        )
        send_button.pack(
            side=tk.LEFT,
            padx=5,
            pady=10
        )

        self.show_menu()

        self.user_input.bind(
            "<Return>",
            lambda event: self.send_message()
        )

    def show_menu(self):

        self.chat_area.insert(
            tk.END,
            "Bot: Welcome to the College Admission Assistant!\n\n"
            "Please select an option:\n"
            "1. View Departments / Courses\n"
            "2. View Vacant Seats\n"
            "3. Take Admission in Computer\n"
            "4. Take Admission in IT\n"
            "5. Exit / Thank You\n\n"
        )

    def send_message(self):

        message = self.sensor.get_input()

        if message == "":
            return

        self.chat_area.insert(
            tk.END,
            "You: " + message + "\n"
        )

        if self.agent_type.get() == "Simple Reflex Agent":

            response = self.simple_agent.response(message)

        else:

            response = self.model_agent.response(message)

        self.actuator.display_response(response)

        self.user_input.delete(
            0,
            tk.END
        )
```

#### SENSOR

```python
# The sensor receives the student's input from the environment.
# In this chatbot, the Entry widget acts as the input source.
class AdmissionSensor:

    def __init__(self, chatbot):
        self.chatbot = chatbot

    def get_input(self):
        return self.chatbot.user_input.get().strip()
```

#### ACTUATOR

```python
# The actuator displays the agent's response to the student.
class AdmissionActuator:

    def __init__(self, chatbot):
        self.chatbot = chatbot

    def display_response(self, response):
        self.chatbot.chat_area.insert(
            tk.END,
            "Bot: " + response + "\n\n"
        )
        self.chatbot.chat_area.see(tk.END)
```

#### SIMPLE REFLEX AGENT

```python
# SIMPLE REFLEX AGENT
# Input → Rule → Response
# It only considers the current input.
# It does NOT maintain or update seat information.

class SimpleReflexAgent:

    def response(self, user_input):

        if user_input == "1":
            return "We offer Computer and IT departments."

        elif user_input == "2":
            return (
                "Current Vacant Seats:\n"
                "Computer Engineering : 120\n"
                "IT Engineering       : 120"
            )

        elif user_input == "3":
            return "Admission confirmed in Computer Engineering."

        elif user_input == "4":
            return "Admission confirmed in IT Engineering."

        elif user_input == "5":
            return "Thank you! All the best for your admission."

        else:
            return "Please select a valid option from the menu."
```

#### MODEL BASED AGENT

```python
# Model-Based Agent:
#
# Input + Internal State
#          ↓
#        Rule
#          ↓
# Response + Updated State
# It maintains the current vacant seats as internal state.

class ModelBasedAgent:

    def __init__(self):

        self.vacant_seats = {
            "Computer": 120,
            "IT": 120
        }

    def response(self, user_input):

        if user_input == "1":

            return "We offer Computer and IT departments."

        elif user_input == "2":

            return (
                "Current Vacant Seats:\n"
                f"Computer Engineering : "
                f"{self.vacant_seats['Computer']}\n"
                f"IT Engineering       : "
                f"{self.vacant_seats['IT']}"
            )

        elif user_input == "3":

            if self.vacant_seats["Computer"] > 0:

                self.vacant_seats["Computer"] -= 1

                return (
                    "Admission confirmed in Computer Engineering.\n"
                    f"Remaining seats: "
                    f"{self.vacant_seats['Computer']}"
                )

            return "Sorry, no Computer seats are available."

        elif user_input == "4":

            if self.vacant_seats["IT"] > 0:

                self.vacant_seats["IT"] -= 1

                return (
                    "Admission confirmed in IT Engineering.\n"
                    f"Remaining seats: "
                    f"{self.vacant_seats['IT']}"
                )

            return "Sorry, no IT seats are available."

        elif user_input == "5":

            return "Thank you! All the best for your admission."

        else:

            return "Please select a valid option from the menu."
```

# Conclusion:
In the Simple Reflex Agent, vacant seats are fixed at 120 and do not change after admission.
For example, even after a student takes admission in Computer Engineering, it still shows 120 seats.

In the Model-Based Agent, vacant seats are stored as internal state and updated after admission.
For example, Computer seats change from 120 to 119 after one admission.

Therefore, the Simple Reflex Agent only reacts to the current input,
while the Model-Based Agent remembers the current seat availability and updates it accordingly.
